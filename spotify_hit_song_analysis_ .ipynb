## 4. Clean and Prepare the Data

The raw file has two problems: streams and Deezer playlists loaded as text instead of numbers, and one row is incomplete. I convert those columns, drop rows missing a model variable, and keep only
songs with real stream counts. I also build the variables the analysis needs, log transforms to handle
skew, season from release month, and performance tiers, so every later section pulls from one clean table.

```python
# ---Prepare the Dataset ---
spotify = spotify_raw.copy()

# streams and Deezer playlists loaded as text — convert to numeric
for col in ['streams','in_deezer_playlists']:
    spotify[col] = to_num(spotify[col])

# drop rows missing a model variable; keep only songs with streams
spotify = spotify.dropna(subset=['streams','in_spotify_playlists',
    'in_apple_playlists','in_deezer_playlists']).reset_index(drop=True)
spotify = spotify[spotify['streams'] > 0].reset_index(drop=True)

# derived variables for the analysis
spotify['season']           = spotify['released_month'].astype(int).apply(month_to_season)
spotify['log_streams']      = np.log1p(spotify['streams'])
spotify['log_spotify']      = np.log1p(spotify['in_spotify_playlists'])
spotify['log_apple']        = np.log1p(spotify['in_apple_playlists'])
spotify['log_deezer']       = np.log1p(spotify['in_deezer_playlists'])
spotify['performance_tier'] = pd.qcut(spotify['streams'], q=4, labels=['Lower','Mid-Lower','Mid-Upper','Top'])
spotify['is_top_tier']      = spotify['performance_tier'] == 'Top'

print('=' * 60)
print('DATA CLEANING SUMMARY')
print('=' * 60)

# what was wrong in the raw data
print(f'Raw dataset:   {spotify_raw.shape[0]:,} songs, {spotify_raw.shape[1]} columns')
print('Fixed 2 columns loaded as text -> numeric: streams, in_deezer_playlists')

# rows removed and why
dropped = spotify_raw.shape[0] - spotify.shape[0]
print(f'Removed {dropped} row(s): missing a model variable or zero streams')

# what the clean data looks like now
print(f'Clean dataset: {spotify.shape[0]:,} songs, 0 missing in model columns')
print('Added derived variables: season, log_streams, performance_tier, is_top_tier')

# preview the cleaned data
print('\nFirst few rows (cleaned):')
display(spotify.head())

# confirm the model variables are now real numbers
print('\nKey model variables (cleaned):')
display(spotify[['streams','in_spotify_playlists','in_apple_playlists','in_deezer_playlists']].describe())
```

## 6. Which Platform's Playlists Predict Streams?

I regress log streams on the logged playlist counts for all three platforms at once, so each platform's
effect is measured while holding the others constant. Logging both sides makes the relationship
interpretable and controls the multicollinearity that raw counts would introduce (confirmed by the VIF
check). Standardized coefficients then put the platforms on the same scale, revealing which playlist
presence carries the real weight and which fades once the others are accounted for.

```python
model_df = spotify[['log_streams','log_spotify','log_apple','log_deezer']].dropna().copy()

predictors = ['log_spotify','log_apple','log_deezer']
X = sm.add_constant(model_df[predictors])
y = model_df['log_streams']

# robust standard errors (HC3) — residuals are heavily skewed, so plain SEs understate uncertainty
model = sm.OLS(y, X).fit(cov_type='HC3')
print(model.summary())

# multicollinearity check — confirms the logged predictors keep VIFs low
vif = pd.DataFrame({
    'variable': predictors,
    'VIF': [variance_inflation_factor(X.values, i + 1) for i in range(len(predictors))]
})
print('\nVariance Inflation Factors:')
print(vif.to_string(index=False))

# standardized coefficients for a fair cross-platform comparison
scaled_df = model_df.copy()
for col in scaled_df.columns:
    scaled_df[col] = (scaled_df[col] - scaled_df[col].mean()) / scaled_df[col].std()
X_s = sm.add_constant(scaled_df[predictors])
scaled_model = sm.OLS(scaled_df['log_streams'], X_s).fit(cov_type='HC3')

coef_summary = pd.DataFrame({
    'platform':          ['Spotify','Apple Music','Deezer'],
    'standardized_coef': [scaled_model.params['log_spotify'],
                          scaled_model.params['log_apple'],
                          scaled_model.params['log_deezer']],
    'p_value':           [model.pvalues['log_spotify'],
                          model.pvalues['log_apple'],
                          model.pvalues['log_deezer']]
}).sort_values('standardized_coef', ascending=False)

# store predictions back on the original stream scale
spotify['predicted_log_streams'] = np.nan
spotify.loc[model_df.index, 'predicted_log_streams'] = model.predict(X)
spotify['predicted_streams'] = np.expm1(spotify['predicted_log_streams'])

print(f'\nR-squared: {model.rsquared:.4f} | Adj R-squared: {model.rsquared_adj:.4f}')
print('\nStandardized coefficients (with significance):')
print(coef_summary.to_string(index=False))
```

**Takeaway:** Playlist placement explains ~62.5% of the variation in streams. Spotify dominates
(standardized 0.66), Deezer is small but significant (0.11, p = 0.010), and Apple is not significant
(0.05, p = 0.23) — once Spotify and Deezer are accounted for, Apple adds no measurable predictive power.

### Robustness check: controlling for confounders

Three things could drive both playlist placement and streams, making Spotify's effect look bigger than it
is: **song age** (streams are cumulative, so older songs have had more time to build up both streams and
playlist adds), **artist fame** (famous artists get playlisted more and streamed more), and **collaboration**
(multi-artist songs may get cross-promoted). I re-run the regression adding all three as controls, a proxy
for artist prominence (how many songs the artist has in the dataset), song age, and artist count, and compare
the coefficients. If Spotify's effect holds steady, the finding isn't just an artifact of these factors.

```python
# --- Robustness check: does the playlist finding survive controlling for confounders? ---
# build the three controls right here so the cell runs no matter what came before
spotify['song_age']          = 2024 - spotify['released_year']
spotify['log_age']           = np.log1p(spotify['song_age'])             # time on platform (proxy)
artist_song_count            = spotify['artist(s)_name'].value_counts()
spotify['artist_prominence'] = spotify['artist(s)_name'].map(artist_song_count)
spotify['log_prominence']    = np.log1p(spotify['artist_prominence'])    # artist fame proxy
# artist_count (collaboration) already exists in the raw data

rc_df = spotify[['log_streams','log_spotify','log_apple','log_deezer',
                 'log_age','log_prominence','artist_count']].dropna().copy()

# Model 1: playlists only
X1 = sm.add_constant(rc_df[['log_spotify','log_apple','log_deezer']])
m1 = sm.OLS(rc_df['log_streams'], X1).fit(cov_type='HC3')

# Model 2: playlists + age, prominence, collaboration
X2 = sm.add_constant(rc_df[['log_spotify','log_apple','log_deezer',
                            'log_age','log_prominence','artist_count']])
m2 = sm.OLS(rc_df['log_streams'], X2).fit(cov_type='HC3')

compare = pd.DataFrame({
    'Model 1 (playlists only)':            m1.params,
    'Model 2 (+ age, prominence, collab)': m2.params
}).round(4)
print(compare)
print(f'\nR-squared: {m1.rsquared:.3f} vs {m2.rsquared:.3f}')
```

**Takeaway:** Spotify's effect barely moved (0.57 → 0.56) after controlling for song age, artist prominence,
and collaboration, so the finding isn't an artifact of older songs, famous artists, or big collaborations.
Notably, prominence and collaboration turn slightly negative once playlists are in the model — artist fame
appears to work *through* playlist placement rather than on top of it.

### Validation: predicting on held-out songs

Split the songs into train and test sets, fit on the training data, and check whether the model predicts about as well on songs it never saw.

```python
# --- Train/test validation: does the model predict on unseen songs? ---
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

X_tt = model_df[predictors]
y_tt = model_df['log_streams']

X_train, X_test, y_train, y_test = train_test_split(
    X_tt, y_tt, test_size=0.2, random_state=42
)

# fit on the training set only, then predict on the held-out test set
tt_model   = sm.OLS(y_train, sm.add_constant(X_train)).fit()
test_preds = tt_model.predict(sm.add_constant(X_test))

print('Train/Test Validation')
print(f'Train R²: {tt_model.rsquared:.3f}')
print(f'Test  R²: {r2_score(y_test, test_preds):.3f}')
```

**Takeaway:** The model achieves a test-set R² of 0.60, nearly identical to the training R² of 0.63 — so it generalizes to unseen songs rather than overfitting the data it was trained on.

## 7. Do a Song's Own Characteristics Predict Success?

Beyond playlists, does what a song actually sounds like separate the biggest hits from the rest?
I check the audio features two ways:

- **First pass (t-tests):** compare each audio trait (danceability, energy, etc.) between top-tier songs
  (top 25% by streams) and the rest, with a Bonferroni correction since I test 8 at once.
- **Validation (logistic regression):** check whether those differences hold up when all features are
  considered together, controlling for their correlations, so I can tell a real signal from a borrowed one.

### Audio features: do top-tier songs sound different?

For each audio trait, I compare the average of top-tier songs (top 25% by streams) against everyone else
using a Welch's t-test (which doesn't assume the two groups have equal variance). Since I test 8 features
at once, I apply a Bonferroni correction, tightening the significance threshold to 0.05 / 8 = 0.0063, so a
feature only counts if it clears that stricter bar. I also report Cohen's d (effect size), because with
~950 songs even tiny differences can be statistically significant, so the effect size shows whether a
real gap is actually large or trivial. This is the first pass, checking each feature on its own.

```python
print('=' * 70)
print('T-TESTS: Do audio features differ between top-tier songs and the rest?')
print('=' * 70)

audio_features = ['danceability_%','valence_%','energy_%','acousticness_%',
                  'instrumentalness_%','liveness_%','speechiness_%','bpm']
audio_features = [c for c in audio_features if c in spotify.columns]
print('Testing features:', audio_features)

# Bonferroni correction: many tests inflate the false-positive rate
alpha_corrected = 0.05 / len(audio_features)
print(f'\nBonferroni-corrected threshold: {alpha_corrected:.4f}\n')

# Cohen's d: effect size (how BIG the gap is, not just whether it's significant)
def cohens_d(a, b):
    n1, n2 = len(a), len(b)
    s1, s2 = a.std(ddof=1), b.std(ddof=1)
    pooled_sd = np.sqrt(((n1 - 1) * s1**2 + (n2 - 1) * s2**2) / (n1 + n2 - 2))
    return (a.mean() - b.mean()) / pooled_sd

results = []
for feature in audio_features:
    top_vals  = spotify[spotify['is_top_tier']][feature].dropna()
    rest_vals = spotify[~spotify['is_top_tier']][feature].dropna()
    t_stat, p_val = ttest_ind(top_vals, rest_vals, equal_var=False)   # Welch's t-test
    results.append({
        'feature':     feature,
        'top_mean':    round(top_vals.mean(), 2),
        'rest_mean':   round(rest_vals.mean(), 2),
        't':           round(t_stat, 4),
        'p':           round(p_val, 6),
        'cohens_d':    round(cohens_d(top_vals, rest_vals), 3),   # effect size
        'significant': p_val < alpha_corrected
    })

print(pd.DataFrame(results).sort_values('p').to_string(index=False))
```

**Takeaway:** Only speechiness and danceability differ significantly between top-tier songs and the rest,
and both are small effects (Cohen's d = −0.28 and −0.22); everything else is negligible. With ~950 songs
the tests detect tiny gaps, so the effect sizes confirm audio barely separates hits from the rest.

### Do the audio differences hold up together?

The t-tests looked at each feature in isolation. Here I predict whether a song is top-tier from all the
audio features at once using logistic regression. Because it controls for the correlations between features,
each coefficient shows a feature's *unique* contribution, revealing which t-test findings are genuinely
independent signals and which fade once the other features are accounted for.

```python
# Logistic regression: which audio features predict being a top-tier song,
# controlling for all the others at once? (the "do they hold up together?" check)
audio_features = ['danceability_%','valence_%','energy_%','acousticness_%',
                  'instrumentalness_%','liveness_%','speechiness_%','bpm']
audio_features = [c for c in audio_features if c in spotify.columns]

logit_df = spotify[audio_features + ['is_top_tier']].dropna().copy()

# standardize features so coefficients are directly comparable across features
X = pd.DataFrame(StandardScaler().fit_transform(logit_df[audio_features]),
                 columns=audio_features, index=logit_df.index)
X = sm.add_constant(X)
y = logit_df['is_top_tier'].astype(int)

logit_model = sm.Logit(y, X).fit(disp=False)
print(logit_model.summary())

# tidy table: coefficient, odds ratio, and p-value, sorted by significance
logit_results = pd.DataFrame({
    'feature':    X.columns,
    'coef':       logit_model.params.values,
    'odds_ratio': np.exp(logit_model.params.values),
    'p_value':    logit_model.pvalues.values
}).query("feature != 'const'").sort_values('p_value')

print('\nStandardized logistic regression — predicting top-tier:')
print(logit_results.to_string(index=False))
```

**Takeaway:** Controlling for all audio features together, only speechiness stays significant (p = 0.001);
danceability fades (p = 0.046), meaning its t-test result was partly borrowed from correlated features.
Audio explains almost none of hit status (pseudo R² ≈ 0.02) — speechiness is the one durable signal.

### Robustness: does the finding depend on the 25% cutoff?

My "top-tier" split uses the top 25% by streams, which is a common but arbitrary choice. To check the
audio findings aren't an artifact of that threshold, I re-run the t-tests for speechiness and danceability
at the top 10%, 25%, and 33%. If a feature stays significant across all three, it's stable to the cutoff.

```python
# --- Robustness: do the audio findings hold at different top-tier cutoffs? ---
for feature in ['speechiness_%', 'danceability_%']:
    print(f'\n{feature}: top vs rest across cutoffs')
    for pct in [0.10, 0.25, 0.33]:
        thr  = spotify['streams'].quantile(1 - pct)   # top `pct` by streams
        top  = spotify[spotify['streams'] >= thr][feature].dropna()
        rest = spotify[spotify['streams'] <  thr][feature].dropna()
        t, p = ttest_ind(top, rest, equal_var=False)
        print(f'  Top {int(pct*100):>2}%: t={t:6.2f}, p={p:.5f}, significant(Bonferroni)={p < 0.0063}')
```

**Takeaway:** Speechiness stays significant at every cutoff (10%, 25%, 33%), so it's robust to the
threshold choice. Danceability only clears the bar at 25% and fails at 10% and 33%, confirming it's a
fragile, cutoff-dependent signal.

### The decisive test: does sound add anything on top of playlists?

I tested playlists and audio in separate frameworks, but the cleanest single test is to put them in one
model and ask whether audio features explain any streaming variation *once playlist placement is
controlled for*. I compare a playlists-only model to a playlists-plus-audio model. If audio barely moves
R² and its coefficients are mostly non-significant, then a song's sound adds little beyond its distribution.

```python
# --- Combined model: does sound add anything on top of playlist placement? ---
audio_features = ['danceability_%','valence_%','energy_%','acousticness_%',
                  'instrumentalness_%','liveness_%','speechiness_%','bpm']
audio_features = [c for c in audio_features if c in spotify.columns]

cm = spotify[['log_streams','log_spotify','log_apple','log_deezer'] + audio_features].dropna().copy()
y = cm['log_streams']

# Model A: playlists only
Xa = sm.add_constant(cm[['log_spotify','log_apple','log_deezer']])
mA = sm.OLS(y, Xa).fit(cov_type='HC3')

# Model B: playlists + audio features
Xb = sm.add_constant(cm[['log_spotify','log_apple','log_deezer'] + audio_features])
mB = sm.OLS(y, Xb).fit(cov_type='HC3')

print(f'R-squared, playlists only:     {mA.rsquared:.4f}')
print(f'R-squared, playlists + audio:  {mB.rsquared:.4f}')
print(f'Gain from adding audio:        {mB.rsquared - mA.rsquared:.4f}')
print('\nAudio coefficients in the combined model (p-values):')
print(mB.pvalues[audio_features].round(4))
```

**Takeaway:** Adding all eight audio features raised R² by just 0.005 (0.625 → 0.630), and only speechiness
stayed significant (p = 0.002). A song's sound adds almost nothing once playlist placement is controlled
for — the clearest single-model confirmation that distribution beats sound.

## 8. Does Release Timing Matter?

Release timing isn't a property of the song itself, it's a business decision, so I treat it as its own question.
Both variables are categorical (season and the four performance tiers), so I use a chi-square test of
independence to check whether they're related, plus Cramér's V to measure how strong any association is.

```python
print('=' * 70)
print('CHI-SQUARE: Is release season related to performance tier?')
print('=' * 70)

# both variables are categorical, so cross-tabulate counts and test for independence
contingency = pd.crosstab(spotify['season'], spotify['performance_tier'])
chi2_stat, chi2_p, chi2_dof, expected = chi2_contingency(contingency)   # keep 'expected' this time

# Cramér's V: how STRONG the association is (0 = none, 1 = perfect)
n = contingency.values.sum()
cramers_v = np.sqrt(chi2_stat / (n * (min(contingency.shape) - 1)))

print(f'chi2={chi2_stat:.4f} | df={chi2_dof} | p={chi2_p:.6f} | Significant: {"Yes" if chi2_p < 0.05 else "No"}')
print(f"Cramer's V (effect size): {cramers_v:.3f}")
print('Min expected count:', round(expected.min(), 1), '(needs to be >= 5 for the test to be valid)')

print('\nContingency table (songs per season x tier):')
print(contingency)

# standardized residuals: which season x tier cells deviate most from "no relationship"?
residuals = (contingency - expected) / np.sqrt(expected)
print('\nStandardized residuals (|value| > 2 = notable):')
print(residuals.round(2))
```

**Takeaway:** Release season and performance are significantly related (χ² = 55.40, p < .001) but the
effect is weak (Cramér's V = 0.14). The residuals show it's driven mainly by Fall — which strongly avoids low performers (−4.46) and skews toward mid-upper songs (+3.15) — plus Spring and Summer over-indexing the bottom tier.

## 9. Do Songs Fall Into Natural Groups?

I use K-means to group songs by their playlist footprint, letting the structure emerge on its own. The
silhouette score peaked sharply at k=2 (0.78) and the elbow agreed, so I use two clusters rather than
forcing four. PCA compresses the three playlist features to 2D so the groups can be plotted.

### How many clusters?

I check k against the data with two standard diagnostics: the elbow (where adding clusters stops reducing
within-group variance) and the silhouette score (how cleanly separated the clusters are). Both point to k=2.

```python
# --- How many clusters? Elbow + silhouette to justify k ---
cluster_features = ['in_spotify_playlists','in_apple_playlists','in_deezer_playlists']
Xk = StandardScaler().fit_transform(spotify[cluster_features].dropna())

Ks = range(2, 9)
inertias, sils = [], []
for k in Ks:
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(Xk)
    inertias.append(km.inertia_)
    sils.append(silhouette_score(Xk, km.labels_))

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(list(Ks), inertias, 'o-', color=LIME); ax1.set_title('Elbow: inertia vs k')
ax1.set_xlabel('k'); ax1.set_ylabel('Inertia')
ax2.plot(list(Ks), sils, 'o-', color=LIME); ax2.set_title('Silhouette vs k')
ax2.set_xlabel('k'); ax2.set_ylabel('Silhouette score')
for ax in (ax1, ax2): apply_style(ax); ax.grid(False)
save_plot('cluster_k_selection.png')
print('Silhouette by k:', {k: round(s, 3) for k, s in zip(Ks, sils)})
```

**Takeaway:** The silhouette score is highest at k = 2 (0.78) and the elbow agrees, so the data points to
two natural groups. I use k = 2 rather than forcing more.

### Building the two clusters

With k=2 confirmed, I group songs by their playlist footprint, name each cluster by its median streams
(lower = Low Exposure, higher = High Exposure), and reduce the three features to 2D with PCA for plotting.

```python
# cluster on playlist footprint (exposure), then describe each group by streams
cluster_features = ['in_spotify_playlists','in_apple_playlists','in_deezer_playlists']
cluster_df       = spotify[cluster_features + ['streams']].dropna().copy()
scaled_features  = StandardScaler().fit_transform(cluster_df[cluster_features])

# k=2: silhouette (0.78) and the heavy imbalance both point to two natural groups
cluster_df['cluster'] = KMeans(n_clusters=2, random_state=42, n_init=10).fit_predict(scaled_features)

# name the two clusters by median streams: lower -> higher
cluster_medians   = cluster_df.groupby('cluster')['streams'].median().sort_values()
group_names       = ['Low Exposure', 'High Exposure']
cluster_label_map = {cid: group_names[i] for i, (cid, _) in enumerate(cluster_medians.items())}
cluster_df['cluster_label'] = cluster_df['cluster'].map(cluster_label_map)

# map labels back onto the main dataframe
spotify['cluster']       = np.nan
spotify['cluster_label'] = np.nan
spotify.loc[cluster_df.index, 'cluster']       = cluster_df['cluster']
spotify.loc[cluster_df.index, 'cluster_label'] = cluster_df['cluster_label']

# reduce the 3 playlist features to 2 dimensions for plotting
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_features)
cluster_df['pca1'] = pca_result[:, 0]
cluster_df['pca2'] = pca_result[:, 1]

print(cluster_df['cluster_label'].value_counts())
print(f'PCA explained variance (2 components): {pca.explained_variance_ratio_.sum():.2%}')
```

**Takeaway:** Songs split into a large **Low Exposure** group (883 songs, ~93%) and a small **High Exposure**
group (69 songs, ~7%) that streams far more. Playlist exposure is highly concentrated, echoing the skew in
the streams distribution.

"map" that squeezes the three playlist measures into two directions so both groups can be seen at once.

## Key Findings

- **Playlist placement is the strongest driver of streaming performance**, explaining ~62.5% of the variation in streams (log-log regression, R² = 0.625). The effect held after controlling for song age, artist prominence, and collaboration.
- **The model validates out-of-sample** — a held-out test set gave R² = 0.60 vs. 0.63 in training, so it predicts unseen songs about as well, confirming a real pattern rather than overfitting.
- **It's overwhelmingly a Spotify story.** Standardized effects: Spotify 0.66 (dominant, significant), Deezer 0.11 (small, positive, significant), Apple Music 0.05 (not significant). Artist fame appears to work *through* playlist placement rather than on top of it.
- **A song's sound barely predicts performance.** Only speechiness holds up as a reliable signal — robust across cutoffs and in a combined model with playlists (p = 0.002) — while every other audio feature, including danceability, fails to hold. Adding all audio features to the playlist model raises R² by just 0.005.
- **Release timing has a weak effect** (χ² = 55.40, p < .001, Cramér's V = 0.14). Winter releases over-index the top tier (30.5% vs ~21% in spring), but the association is driven mainly by Fall (avoiding the bottom tier, skewing mid-upper) and by Spring and Summer over-indexing the bottom tier.
- **Playlist exposure is highly concentrated.** Songs split into a large low-exposure majority (~93%) and a small high-exposure group (~7%) that streams far more.

**Bottom line:** streaming success is driven far more by *where* a song is distributed than by *what* it sounds like.

## Implications

*(These are suggestions based on associations, not proven causes — see Limitations.)*

- **Prioritize Spotify playlist pitching.** It's by far the strongest predictor; Apple shows no significant independent effect and Deezer only a small one, and artist fame appears to work through playlist placement, making Spotify pitching the highest-leverage move.
- **Don't over-invest in engineering a song's "sound."** Audio features barely relate to performance — only speechiness shows a small link, and even that adds almost nothing once distribution is accounted for.
- **Treat release timing as a minor factor.** Fall and Winter releases skew slightly higher, but the effect is weak (Cramér's V = 0.14), so it's a light consideration, not a core strategy.

## Limitations

- **Correlation, not causation.** Playlists and streams reinforce each other (a hit gets added to more playlists, which drives more streams), so these are strong predictors, not proven causes.
- **Selection bias.** The data covers only the top songs of 2023, so findings describe what separates big hits from bigger ones, not what makes an unknown song succeed.
- **Unmeasured confounders.** Song quality, marketing spend, and social-media virality all plausibly drive both playlist placement and streams, and none are in the dataset.
- **Non-independent observations.** Some artists appear multiple times, so their songs aren't fully independent; clustering standard errors by artist would be the rigorous fix.

## Next Steps

- **Test beyond the hits.** Re-run the playlist–streams model on a broader sample that includes non-charting songs, to check whether the pattern holds outside the top-song bubble.
- **Add external signals.** Bring in artist followers, monthly listeners, or marketing data to control for fame and promotion directly rather than with a proxy.
- **Move toward causation.** Use longitudinal data (streams and playlist adds over time) to untangle the playlist↔streams feedback loop.
- **Tighten the inference.** Cluster standard errors by artist to account for repeated artists — the main remaining refinement, since the model already validates on a held-out test set (test R² 0.60 vs. 0.63 train).
